```python
import pandas as pd
```

```python
df = pd.read_csv("../dataset_report_anonimizzati_llama_cpp.csv")
```

```python
df
```

```python
lista_testi_duplicati = df[df["testo_originale"].duplicated()]["testo_originale"].unique().tolist()
```

```python
for testo, gruppo in zip(lista_testi_duplicati, range(len(lista_testi_duplicati))):
    df.loc[df["testo_originale"] == testo, "gruppo"] = gruppo
```

```python
df_duplicati = df[df["gruppo"].notnull()]
df_duplicati
```

```python
df_duplicati = df_duplicati.drop(columns=["report_anonimo"])
df_duplicati
```

```python
df_duplicati.to_csv("df_duplicati.csv")
```
