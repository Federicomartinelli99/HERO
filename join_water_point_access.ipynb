#### Aggregazione per area geografica (admin2) e per data (cumulativa per report_date)

```python
# ── 1. Tutti i checkpoint (zone × date distinte) ────────────────────────────
checkpoints = df_water[
    ['country_id', 'adm1_pcode_manual', 'adm2_pcode_manual', 'report_date']
].drop_duplicates()

# ── 2. Attributi per punto d'acqua ──────────────────────────────────────────
points = df_water

# ── 3. Per ogni checkpoint, tutti i punti con report_date <= checkpoint ──────
merged = checkpoints.merge(
    points,
    on=['country_id', 'adm1_pcode_manual', 'adm2_pcode_manual'],
    suffixes=('_checkpoint', '_point')
)
merged = merged[merged['report_date_point'] <= merged['report_date_checkpoint']]

# ── 4. Aggregazione cumulativa di tutto insieme ──────────────────────────────
df_water_agg = (
    merged
    .groupby(['country_id', 'adm1_pcode_manual', 'adm2_pcode_manual', 'report_date_checkpoint'])
    .agg(
        local_population=('local_population', 'sum'),
        assigned_population=('assigned_population', 'sum'),
        water_points_count=('wpdx_id', 'nunique'),
        criticality = ('criticality','mean'),
        functional_points_count=('wpdx_id', lambda x: x[merged.loc[x.index, 'functional_binary'] == 'Functional'].nunique()),

    )
    .reset_index()
    .rename(columns={'report_date_checkpoint': 'report_date'})
)
```

```python
url_semiwide = "https://raw.githubusercontent.com/Federicomartinelli99/HERO/refs/heads/main/ipc/ipc_global_area_wide_pcoded.csv"
df_ipc = pd.read_csv(url_semiwide, parse_dates = ['From', 'To'], dayfirst = True)
```

```python
country_match = set(df_ipc['Country'].unique()) & set(df_water['country_id'].unique())
```

#### Join con dataset ipc a livello admin2

```python
import numpy as np


df_water_agg['report_date'] = pd.to_datetime(df_water_agg['report_date'])

# 1. Creiamo il set univoco di intervalli IPC a livello Admin 2
df_ipc_adm2 = df_ipc[['Country', 'From', 'To', 'adm1_pcode', 'adm2_pcode']].drop_duplicates().reset_index(drop=True)
df_ipc_adm2['adm2_row_id'] = df_ipc_adm2.index

# 2. Merge Geografico (Left Join)
df_geo_merged = df_ipc_adm2.merge(
    df_water_agg,
    left_on=['Country', 'adm1_pcode', 'adm2_pcode'],
    right_on=['country_id', 'adm1_pcode_manual', 'adm2_pcode_manual'],
    how='left'
)

# 3. Calcolo della distanza temporale
def calculate_distance(df):
    mask_inside = (df['report_date'] >= df['From']) & (df['report_date'] <= df['To'])
    dist_from = (df['report_date'] - df['From']).dt.days.abs()
    dist_to = (df['report_date'] - df['To']).dt.days.abs()
    return np.where(mask_inside, 0, np.minimum(dist_from, dist_to))

df_geo_merged['distanza_giorni'] = calculate_distance(df_geo_merged)

# 4. Selezione per ogni intervallo Admin 2 (adm2_row_id)

# A. Match perfetti (distanza 0): teniamo il checkpoint più recente nell'intervallo
mask_perfect = df_geo_merged['distanza_giorni'] == 0
df_perfect_candidates = df_geo_merged[mask_perfect].sort_values(by=['adm2_row_id', 'report_date'])
df_perfect_avg = df_perfect_candidates.drop_duplicates('adm2_row_id', keep='last')[
    ['adm2_row_id', 'local_population', 'assigned_population', 'water_points_count',
     'functional_points_count', 'criticality', 'report_date', 'distanza_giorni']
]

# B. Recupero per prossimità (Orfani): prendiamo il più vicino solo se non c'è un match perfetto
matched_ids = df_perfect_avg['adm2_row_id'].unique()
df_orphans = df_geo_merged[
    (~df_geo_merged['adm2_row_id'].isin(matched_ids)) &
    (df_geo_merged['water_points_count'].notna())
].copy()

df_orphans_closest = df_orphans.sort_values(by=['adm2_row_id', 'distanza_giorni']).drop_duplicates('adm2_row_id', keep='first')

# C. Unione dei risultati mappati agli ID univoci di df_ipc_adm2
df_idp_final_map = pd.concat([
    df_perfect_avg,
    df_orphans_closest[['adm2_row_id', 'local_population', 'assigned_population', 'water_points_count',
    'functional_points_count', 'criticality', 'report_date', 'distanza_giorni']]
])

# 5. Integrazione finale nel dataset originale
df_ipc_adm2_integrated = df_ipc_adm2.merge(
    df_idp_final_map[['adm2_row_id', 'local_population', 'assigned_population', 'water_points_count',
    'functional_points_count', 'criticality', 'report_date', 'distanza_giorni']],
    on='adm2_row_id',
    how='left'
)

df_final_integrated = df_ipc.merge(
    df_ipc_adm2_integrated[['Country', 'From', 'To', 'adm1_pcode', 'adm2_pcode', 'local_population', 'assigned_population', 'water_points_count',
    'functional_points_count', 'criticality', 'report_date', 'distanza_giorni']],
    on=['Country', 'From', 'To', 'adm1_pcode','adm2_pcode'],
    how='left'
)

# --- VERIFICA ---
print(f"Verifica righe: {len(df_final_integrated)} (Originale IPC: {len(df_ipc)})")
print(f"Righe con dato acqua integrato: {df_final_integrated['water_points_count'].notna().sum()}")
```

##### Copertura dato punti accesso acqua per paese

```python
coverage_by_country = (
    df_final_integrated
    .groupby('Country')['water_points_count']
    .agg(
        righe_totali='size',
        righe_con_acqua='count'  # 'count' su una Series ignora i NaN
    )
)
coverage_by_country['pct_coperto'] = (
    coverage_by_country['righe_con_acqua'] / coverage_by_country['righe_totali']
)
coverage_by_country = coverage_by_country.sort_values('pct_coperto', ascending=False)

print(coverage_by_country)
```
